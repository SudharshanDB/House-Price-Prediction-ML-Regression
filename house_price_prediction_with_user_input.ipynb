```python
import pandas as pd
import joblib

# 1. Load saved assets
trained_models_list = joblib.load('trained_models_list.pkl')
scaler = joblib.load('scaler.pkl')
feature_columns = joblib.load('feature_columns.pkl')

print("Choose a model to use for prediction:")
for idx, (name, model) in enumerate(trained_models_list):
    print(f"[{idx}] {name}")

# Let user pick by index
model_index = int(input("\nEnter the number corresponding to your chosen model: "))
chosen_name, chosen_model = trained_models_list[model_index]

print(f"\nUsing model: {chosen_name}")

# 2. Define new house details (Make sure these keys match your training features exactly!)
new_house = {
    'bedrooms': 3,
    'bathrooms': 2.0,
    'sqft_living': 2100,
    'sqft_lot': 6500,
    'floors': 1.0,
    'waterfront': 0,
    'view': 0,
    'condition': 3,
    'sqft_above': 1600,       # Added
    'sqft_basement': 500,     # Added (1600 + 500 = 2100)
    'yr_built': 1998
}

# 3. Convert user input into a DataFrame
input_df = pd.DataFrame([new_house], columns=feature_columns)

# DEBUG CHECK: Print the DataFrame to see if any NaNs appeared
print("\n--- Checking Input DataFrame ---")
print(input_df)
print("\nAny missing values (NaNs)?:", input_df.isna().any().any())

# If there are no NaNs, proceed safely:
if input_df.isna().any().any():
    print("\nError: Your input dictionary has a mismatch with the training feature columns. Check your spelling!")
else:
    input_scaled = scaler.transform(input_df)
    predicted_price = chosen_model.predict(input_scaled)[0]
    print(f"\nPredicted Price: ${2.7 * predicted_price:,.2f}")
```
